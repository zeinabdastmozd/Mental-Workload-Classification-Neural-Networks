```python
def label_to_vector(label):
    label_y = np.ones((2,360))
    for x in range(len(label)) :
        if label[x] == 0 :
           label_y[1][x] = 0
        else:
           label_y[0][x] = 0
    return label_y
#print(label[0].index(0))
y = label_to_vector(label[0]) 
y
```

```python
#using the previous labs function and tryed on eeg data
def gradient_weight(x2, y2, initialization_num, w, b):
    k2 = y2.shape[1]
    wt2 = np.transpose(w)
    
    z2 = np.dot(wt2,x2) + b

    yhat2 = 1 / (1 + np.exp(-z2))

    y_diff=np.subtract(yhat2, y2)
    y_diff_transpose=np.transpose(y_diff)
    gradient_w= (np.dot(x2, y_diff_transpose))/k2
    return gradient_w

def gradient_bias(x2, y2, initialization_num, w, b):
    k2 = y2.shape[1]
    wt2 = np.transpose(w)
    
    z2 = np.dot(wt2,x2) + b

    yhat2 = 1 / (1 + np.exp(-z2))

    y_diff=np.subtract(yhat2, y2)
    gradient_b=np.sum(y_diff)/k2
    return gradient_b


def forward_propagation_loss(X1, initialization_num, y1, w, b):
   
    yhat = prediction(X1, w, b)
    #print(f'Y1 : {y1.shape}, yhat : {yhat.shape}, X1 : {X1.shape}')
    
    
    loss= -np.sum((y1*(np.log(yhat)))+((1-y1)*(np.log(1-yhat))))/y1.shape[1]
    
    return yhat,loss

def model(X1, y1, initialization_num, iterations_num, learning_rate):
    w = np.full((X1.shape[0],y1.shape[0]), initialization_num)
    b = np.full((y1.shape[0], 1), initialization_num)
    
    for i in range (iterations_num):
        yhat, loss = forward_propagation_loss(X1, initialization_num, y1, w, b)
        w = w - learning_rate * gradient_weight(X1, y1, initialization_num, w, b)
        b = b - learning_rate * gradient_bias(X1, y1, initialization_num, w, b)
        #print(f"Iteration {i+1}: loss = {loss}")
        if loss <=0.0001:
            break
            
    return w,b

def prediction(X1, w, b):
     wt = np.transpose(w)
     z = np.dot(wt,X1) + b
     yhat = 1 / (1 + np.exp(-z))
     return yhat




initialization_num = 0
learning_rate = 0.00001
iterations_num =30


w,b = model(X, y, initialization_num, iterations_num, learning_rate)
print(f"Weight : {w}")
print()
print(f"Bias : {b}")
yhat, loss = forward_propagation_loss(X, initialization_num, y, w, b)
print()
print(f"Prediction: {yhat}")
print()
print(f"Loss: {loss}")

#result the loss is decresing and i got 0.8 at learning rate 0000.1
```

```python
# define a funtion to calculate the accuracy
def calculate_accuracy(y_true, y_pred):
    """
    Calculate accuracy.

    Parameters:
    - y_true: True labels (ground truth), shape (samples,)
    - y_pred: Predicted labels, shape (samples,)

    Returns:
    - Accuracy
    """
    print(f'y true : {y_true.shape}, Y pred : {y_pred.shape}')
    accuracies = []
    accuracy = 0
    for i in range(y_true.shape[0]):
        for j in range(y_true.shape[1]):
            yt = y_true[i][j]
            if (yt == 0):
                accuracy = 1 - y_pred[i][j]
            else:
                accuracy = y_pred[i][j]
            accuracies.append(accuracy)
                
    accuracies_np = np.array(accuracies)  
    
    print(f'Accuracies : {accuracies_np}, mean accuracy :{np.mean(accuracies_np)}')
    
    return np.mean(accuracies_np)
```

```python
import numpy as np
from sklearn.ensemble import RandomForestClassifier

def k_fold_cross_validation(k=5):
    """
    Perform K-Fold Cross-Validation.

    Parameters:
    - X: Input features (numpy array or list)
    - y: Target variable (numpy array or list)
    - model: Machine learning model (instance of RandomForestClassifier)
    - k: Number of folds (default is 5)

    Returns:
    - List of tuples, each tuple containing train_indices and test_indices for a fold
    """
        
    fold_size = n_samples // k
    #indices = np.arange(n_samples)
    #np.random.shuffle(indices)

    fold_indices = []
    for i in range(k):
        start = i * fold_size
        end = min((i + 1) * fold_size, n_samples)
        test_indices = indices[start:end]
        train_indices = np.concatenate((indices[:start], indices[end:]))
        fold_indices.append((train_indices, test_indices))

    return fold_indices

def evaluate_model(X, y, k=5):
    """
    Evaluate a model using K-Fold Cross-Validation.

    Parameters:
    - X: Input features (numpy array or list)
    - y: Target variable (numpy array or list)
    - model: Machine learning model (instance of RandomForestClassifier)
    - k: Number of folds (default is 5)

    Returns:
    - List of evaluation metrics for each fold
    """
    fold_indices = k_fold_cross_validation(k)
    evaluation_metrics = []
   
    initialization_num = 0
    learning_rate = 0.0000001
    iterations_num =30
    
    for train_indices, test_indices in fold_indices:
        X_train, y_train = X[:,train_indices], y[:, train_indices]
        X_test, y_test = X[:,test_indices], y[:, test_indices]

        # Train the model
        # model.fit(, y_train.ravel())  # ravel y_train to convert it into a 1D array
        print(f'X train : {X_train.shape}')
        w,b = model(X_train, y_train, initialization_num, iterations_num, learning_rate)

        # Evaluate the model
        yhat, loss = forward_propagation_loss(X_test, initialization_num, y_test, w, b)
        accuracy =calculate_accuracy(y_test, yhat)  # ravel y_test to convert it into a 1D array

        evaluation_metrics.append(accuracy)
       
    return evaluation_metrics



```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models

# Define the CNN model
def create_cnn_model(input_shape, num_classes):
    model = models.Sequential([
        layers.Conv1D(filters=32, kernel_size=3, activation='relu', input_shape=input_shape, name='conv_1'),
        layers.MaxPooling1D(pool_size=2, name='layer2'),
        layers.Conv1D(filters=64, kernel_size=3, activation='relu', name='layer3'),
        layers.MaxPooling1D(pool_size=2, name='layer4'),
        layers.Flatten( name='layer5'),
        layers.Dense(128, activation='relu', name='layer6'),
        layers.Dropout(0.5, name='layer7'),
        layers.Dense(num_classes, activation='softmax', name='layer8')
    ])
    return model
```

```python

from tqdm import tqdm

def evaluate_model_deep(X, y, k=5):
    """
    Evaluate a model using K-Fold Cross-Validation.

    Parameters:
    - X: Input features (numpy array or list)
    - y: Target variable (numpy array or list)
    - model: Machine learning model (instance of RandomForestClassifier)
    - k: Number of folds (default is 5)

    Returns:
    - List of evaluation metrics for each fold
    """
    fold_indices = k_fold_cross_validation(k)
    evaluation_metrics = []
       
    for train_indices, test_indices in fold_indices:
        X_train, y_train = X[:,train_indices], y[:, train_indices]
        X_test, y_test = X[:,test_indices], y[:, test_indices]

        # Train the model
        # model.fit(, y_train.ravel())  # ravel y_train to convert it into a 1D array
        print(f'X train : {X_train.T.shape}')
           
        #X_train_m = X_train.reshape(X_train.shape[0], X_train.shape[1], 1)

        X_train_m = X_train.T
        X_train_m = X_train_m.reshape(X_train_m.shape[0], X_train_m.shape[1], 1)        
        input_shape = X_train_m.shape[1:] 
        #input_shape = X_train.T.shape  # Shape of input data (number of channels, number of time steps)
        num_classes = len(np.unique(y_train))  # Number of classes
        print( input_shape)
        print(num_classes)
       
        #accuracy =calculate_accuracy(y_test.T, y_pred)  # ravel y_test to convert it into a 1D array
        # Create and compile the model
        model = create_cnn_model(input_shape, num_classes)
        #model.summary()
        model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

        X_train = X_train.T
        X_test = X_test.T
        
        # Train the model
        X_train_reshaped = X_train.reshape(X_train.shape[0],  X_train.shape[1])
        X_val_reshaped = X_test.reshape(X_test.shape[0], X_test.shape[1])
        print(X_train_reshaped.shape)
        print(y_train.T.shape)
        model.fit(X_train_reshaped, y_train.T, epochs=10, batch_size=66, validation_data=(X_val_reshaped, y_test.T))

        # Evaluate the model
        test_loss, test_accuracy = model.evaluate(X_val_reshaped, y_test.T)
        print(f'Test loss: {test_loss}, Test accuracy: {test_accuracy}')
        evaluation_metrics.append(test_accuracy)

    return evaluation_metrics
```
